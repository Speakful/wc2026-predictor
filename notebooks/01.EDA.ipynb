- Pre-tournament snapshot to use: `2026-05-27` (updated 2 weeks before kickoff) — filters down to exactly 48 rows, one per team

## Dataset 4 — 2026 WC Match Data

### df_matches_2026
- 104 rows, 8 columns — all fixtures for the 2026 World Cup
- `home_team_id` and `away_team_id` are floats — foreign keys linking to `df_teams_2026`
- `city_id` links to `df_host_cities` — useful for host continent advantage feature
- `stage_id` links to `df_stages_2026` — stage progression logic for simulation
- `kickoff_at` confirms tournament starts June 11, 2026
- `match_label` identifies group (A–L) and knockout stage

### df_teams_2026
- 48 rows, 5 columns — one row per qualified team
- 6 placeholders updated manually with confirmed qualifiers:
  - UEFA Playoff A → Bosnia and Herzegovina (BIH)
  - UEFA Playoff B → Sweden (SWE)
  - UEFA Playoff C → Türkiye (TUR)
  - UEFA Playoff D → Czechia (CZE)
  - FIFA Playoff 1 → DR Congo (COD)
  - FIFA Playoff 2 → Iraq (IRQ)
- `is_placeholder` is now False for all 48 teams

### df_stages_2026
- 7 rows, 3 columns — one row per tournament stage
- Stages in order: Group Stage → Round of 32 → Round of 16 → Quarterfinals → Semifinals → Third Place Playoff → Final
- `stage_order` drives the progression logic in the Monte Carlo simulation engine
- Third Place Playoff is kept — relevant for podium probabilities (1st, 2nd, 3rd)

### df_host_cities
- 16 rows, 6 columns — one row per host city
- 11 USA, 3 Mexico, 2 Canada
- `region_cluster` (East/Central/West) is the key feature — used for host continent advantage
- Columns to drop: `venue_name`, `airport_code`
